```python
import pandas as pd
sales =  pd.read_parquet(r"E:\Finemed_PharmaAI\data\04_silver\sales")

demand_df = pd.read_csv("demand_df.csv")
```

## Lag Features ##

```python
# Lag period
lag_days = [1, 2, 3, 7, 14, 30]

for lag in lag_days:
    demand_df[f"Lag_{lag}"] = (
        demand_df
        .groupby("MDCODE")["Demand_Qty"]
        .shift(lag)
    )
```

```python
# Rolling Mean Features


rolling_windows = [7, 14, 30]

for window in rolling_windows:
    demand_df[f"Rolling_Mean_{window}"] = (
        demand_df
        .groupby("MDCODE")["Demand_Qty"]
        .transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).mean()
        )
    )
```

## Rolling Mean ##

```python
# -------------------------------
# Rolling Mean Features
# -------------------------------

rolling_windows = [7, 14, 30]

for window in rolling_windows:
    demand_df[f"Rolling_Mean_{window}"] = (
        demand_df
        .groupby("MDCODE")["Demand_Qty"]
        .transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).mean()
        )
    )
```

```python
# -------------------------------
# Rolling Standard Deviation
# -------------------------------

for window in rolling_windows:
    demand_df[f"Rolling_STD_{window}"] = (
        demand_df
        .groupby("MDCODE")["Demand_Qty"]
        .transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).std()
        )
    )
```

```python
# -------------------------------
# Rolling Maximum
# -------------------------------

for window in [7, 30]:
    demand_df[f"Rolling_Max_{window}"] = (
        demand_df
        .groupby("MDCODE")["Demand_Qty"]
        .transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).max()
        )
    )
```

```python
# -------------------------------
# Rolling Minimum
# -------------------------------

for window in [7, 30]:
    demand_df[f"Rolling_Min_{window}"] = (
        demand_df
        .groupby("MDCODE")["Demand_Qty"]
        .transform(
            lambda x: x.shift(1).rolling(window=window, min_periods=1).min()
        )
    )
```

## Trend Features ##

```python
# Day-to-day demand difference
demand_df["Demand_Diff_1"] = (
    demand_df
    .groupby("MDCODE")["Demand_Qty"]
    .transform(lambda x: x.diff())
)
```

```python
# Percentage growth from previous day
demand_df["Demand_Growth_Rate"] = (
    demand_df
    .groupby("MDCODE")["Demand_Qty"]
    .transform(lambda x: x.pct_change())
)
```

```python
# Difference compared to last week's demand
demand_df["Weekly_Trend"] = (
    demand_df["Demand_Qty"] - demand_df["Lag_7"]
)
```

## Price Features ##

```python
demand_df["Profit_Margin"] = (
    demand_df["Selling_Rate"] -
    demand_df["Purchase_Rate"]
)
```

```python
demand_df["Profit_Percentage"] = (
    (
        demand_df["Selling_Rate"] -
        demand_df["Purchase_Rate"]
    )
    /
    demand_df["Purchase_Rate"]
) * 100
```

```python
demand_df["Profit_Percentage"] = (
    (
        demand_df["Selling_Rate"] -
        demand_df["Purchase_Rate"]
    )
    /
    demand_df["Purchase_Rate"]
) * 100
```

```python
demand_df["Discount_Percentage"] = (
    (
        demand_df["Maximum_Retail_Price"] -
        demand_df["Selling_Rate"]
    )
    /
    demand_df["Maximum_Retail_Price"]
) * 100
```

```python
demand_df["Price_Change"] = (
    demand_df
    .groupby("MDCODE")["Selling_Rate"]
    .transform(lambda x: x.diff())
)
```

## Product Features ##

```python
### Medicine Frquency 
medicine_frequency = (
    demand_df
    .groupby("MDCODE")
    .size()
)

demand_df["Medicine_Frequency"] = (
    demand_df["MDCODE"]
    .map(medicine_frequency)
)
```

```python
### Average Historical Demand

medicine_avg = (
    demand_df
    .groupby("MDCODE")["Demand_Qty"]
    .mean()
)

demand_df["Medicine_Avg_Demand"] = (
    demand_df["MDCODE"]
    .map(medicine_avg)
)
```

```python
### Demand Volatility 

medicine_std = (
    demand_df
    .groupby("MDCODE")["Demand_Qty"]
    .std()
)

demand_df["Medicine_Demand_STD"] = (
    demand_df["MDCODE"]
    .map(medicine_std)
)
```

```python
### Maximum Historical Demand 

medicine_max = (
    demand_df
    .groupby("MDCODE")["Demand_Qty"]
    .max()
)

demand_df["Medicine_Max_Demand"] = (
    demand_df["MDCODE"]
    .map(medicine_max)
)
```

```python
### Number of Selling Days 

selling_days = (
    demand_df
    .groupby("MDCODE")["INVDT"]
    .nunique()
)

demand_df["Medicine_Selling_Days"] = (
    demand_df["MDCODE"]
    .map(selling_days)
)
```
