```python
import pandas as pd

from IPython.display import display

from ydata_profiling import ProfileReport

sales = pd.read_parquet(r"E:\Finemed_PharmaAI\data\04_silver\sales")

sales_report = ProfileReport(
    sales,
    title = "Sales Report",
    explorative = True
)

display(sales.head().style)
```

```python
columns = [
    # Product
    "MDCODE",
    "MDNAME",
    "PACKG",
    "DETAIL",

    # Time
    "INVDT",
    "day",
    "day_name",
    "day_of_week",
    "week",
    "month",
    "month_name",
    "quarter",
    "year",
    "is_weekend",
    "financial_year",
    "financial_month",

    # Price
    "RATE",
    "PRATE",
    "MRP",

    # Target
    "QTY"
]
```

```python
import pandas as pd

# Select required columns
demand_df = sales[columns].copy()

# Rename for clarity
demand_df = demand_df.rename(
    columns={
        "QTY": "Demand_Qty",
        "RATE": "Selling_Rate",
        "PRATE": "Purchase_Rate",
        "MRP": "Maximum_Retail_Price"
    }
)
```

```python
display(demand_df.head().style)
```

```python
demand_df = (
    demand_df
    .groupby(
        ["MDCODE", "INVDT"],
        as_index=False
    )
    .agg(
        {
            # Target
            "Demand_Qty": "sum",

            # Price
            "Selling_Rate": "mean",
            "Purchase_Rate": "mean",
            "Maximum_Retail_Price": "mean",

            # Product Information
            "MDNAME": "first",
            "PACKG": "first",
            "DETAIL": "first",

            # Calendar Features
            "day": "first",
            "day_name": "first",
            "day_of_week": "first",
            "week": "first",
            "month": "first",
            "month_name": "first",
            "quarter": "first",
            "year": "first",
            "is_weekend": "first",
            "financial_year": "first",
            "financial_month": "first"
        }
    )
)
```
